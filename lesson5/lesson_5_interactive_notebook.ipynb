# Part 3: Activation Functions

Activation functions add non-linearity.

Without activation functions, a deep network would just be a long linear equation.

Common activation functions:

- ReLU
- Sigmoid
- Tanh

## ReLU Activation

ReLU means Rectified Linear Unit.

Formula:

```python
max(0, x)
```

If x is negative, return 0.  
If x is positive, return x.

````python
def relu(x):
    return max(0, x)

print(relu(-5))
print(relu(0))
print(relu(4))
````

## Sigmoid Activation

Sigmoid squashes values between 0 and 1.

It is often used when we want a probability-like output.

````python
import math

def sigmoid(x):
    return 1 / (1 + math.exp(-x))

print(sigmoid(-2))
print(sigmoid(0))
print(sigmoid(2))
````

## Tanh Activation

Tanh squashes values between -1 and 1.

````python
import math

def tanh(x):
    return math.tanh(x)

print(tanh(-2))
print(tanh(-4))
print(tanh(0))
print(tanh(2))
````

# Part 5: Loss Functions

A loss function measures how wrong the prediction is.

Example:

- Prediction = 8
- Actual = 10
- Error = 2

One common loss is Mean Squared Error.

````python
prediction = 8
actual = 10

squared_error = (actual - prediction) ** 2

print("Squared error:", squared_error)
````

## Mean Squared Error

Mean Squared Error, or MSE, averages squared errors over many predictions.

````python
predictions = [8, 6, 9, 4]
actuals = [10, 5, 9, 7]

errors = []

for prediction, actual in zip(predictions, actuals):
    error = (actual - prediction) ** 2
    errors.append(error)

mse = sum(errors) / len(errors)

print("Errors:", errors)
print("MSE:", mse)
````

# Part 6: Gradient Descent

Gradient descent updates weights to reduce loss.

Simplified formula:

```python
new_weight = old_weight - learning_rate * gradient
```

The gradient tells us the direction of error.  
The learning rate controls how big each update step is.

````python
weight = 0.5
learning_rate = 0.1
gradient = -2

new_weight = weight - learning_rate * gradient

print("Old weight:", weight)
print("New weight:", new_weight)
````

## Gradient Descent Simulator

We will train a tiny model:

```python
prediction = input_value * weight
```

The goal is to make the prediction closer to the actual value.

````python
input_value = 2
actual = 10
weight = 1.0
learning_rate = 0.05

for epoch in range(18):
    prediction = input_value * weight
    error = prediction - actual
    loss = error ** 2

    gradient = 2 * error * input_value
    weight = weight - learning_rate * gradient

    print(f"Epoch {epoch+1}: prediction={prediction:.2f}, loss={loss:.2f}, weight={weight:.2f}")
````

# Part 7: Backpropagation

Backpropagation is the process of moving backward from the error to determine which weights caused the mistake.

Beginner-friendly explanation:

1. The network makes a prediction.
2. The loss function measures the error.
3. Backpropagation assigns blame to weights.
4. Gradient descent updates the weights.
5. The model tries again.

## Full Training Loop

The full loop is:

Forward → Loss → Backward → Update → Repeat

````python
# Simplified training loop

input_value = 3
actual = 12
weight = 0.5
learning_rate = 0.05

for epoch in range(15):
    # Forward
    prediction = input_value * weight

    # Loss
    error = prediction - actual
    loss = error ** 2

    # Backward / gradient
    gradient = 2 * error * input_value

    # Update
    weight = weight - learning_rate * gradient

    print(f"Epoch {epoch+1}: prediction={prediction:.2f}, loss={loss:.2f}, weight={weight:.2f}")
````

# Part 8: Recommendation System Training Intuition

A recommendation model tries to predict whether a user will like an item.

Possible features:

- Genre match
- Item popularity
- User history
- Recent feedback
- Rating history

The model makes a prediction, checks how wrong it was, updates weights, and repeats.

````python
def recommendation_score(genre_match, popularity, recent_like, weights, bias):
    score = (
        genre_match * weights[0] +
        popularity * weights[1] +
        recent_like * weights[2] +
        bias
    )
    return score

weights = [0.4, 0.3, 0.3]
bias = -2

score = recommendation_score(
    genre_match=5,
    popularity=4,
    recent_like=5,
    weights=weights,
    bias=bias
)

print("Recommendation score:", score)
````

## Query Tower and Candidate Tower

A **tower** is just a small neural network that turns an ID (user or item) into an embedding vector.

- The **Query Tower** turns a user ID into a user embedding.
- The **Candidate Tower** turns an item ID into an item embedding.

Below is a tiny, toy version of both towers using a handful of fake users and items.

````python
# Toy vocabulary for demonstration purposes
user_ids = ["user_1", "user_2", "user_3"]
item_ids = ["item_a", "item_b", "item_c", "item_d"]

embedding_dimension = 4


class QueryTower(tf.keras.Model):
    """Turns a user ID into a user embedding."""

    def __init__(self, user_ids, embedding_dimension):
        super().__init__()
        self.user_lookup = tf.keras.layers.StringLookup(vocabulary=user_ids, mask_token=None)
        self.user_embedding = tf.keras.layers.Embedding(len(user_ids) + 1, embedding_dimension)

    def call(self, user_id):
        return self.user_embedding(self.user_lookup(user_id))


class CandidateTower(tf.keras.Model):
    """Turns an item ID into an item embedding."""

    def __init__(self, item_ids, embedding_dimension):
        super().__init__()
        self.item_lookup = tf.keras.layers.StringLookup(vocabulary=item_ids, mask_token=None)
        self.item_embedding = tf.keras.layers.Embedding(len(item_ids) + 1, embedding_dimension)

    def call(self, item_id):
        return self.item_embedding(self.item_lookup(item_id))


query_tower = QueryTower(user_ids, embedding_dimension)
candidate_tower = CandidateTower(item_ids, embedding_dimension)

print("Query Tower and Candidate Tower created.")
````

## Retrieval: Finding Likely Candidates

To retrieve candidates for a user, we:

1. Pass the user ID through the Query Tower to get a user embedding.
2. Pass every item ID through the Candidate Tower to get item embeddings.
3. Score each item by how similar its embedding is to the user embedding (dot product).
4. The highest scores are the most likely candidates.

````python
sample_user = tf.constant(["user_1"])
user_embedding = query_tower(sample_user)

item_batch = tf.constant(item_ids)
item_embeddings = candidate_tower(item_batch)

retrieval_scores = tf.linalg.matmul(user_embedding, item_embeddings, transpose_b=True).numpy()[0]

print("Retrieval scores for user_1:")
for item_id, score in zip(item_ids, retrieval_scores):
    print(f"  {item_id}: {score:.4f}")
````

## Ranking: Re-Scoring Candidates

Ranking takes the retrieved candidates and re-scores them with extra metadata (for example, item popularity) to personalize the final order.

````python
# Extra metadata used only during ranking, not retrieval
item_popularity = {"item_a": 0.9, "item_b": 0.4, "item_c": 0.7, "item_d": 0.2}

combined_scores = {
    item_id: score + item_popularity[item_id]
    for item_id, score in zip(item_ids, retrieval_scores)
}

ranked_items = sorted(combined_scores.items(), key=lambda pair: pair[1], reverse=True)

print("Final re-ranked recommendations for user_1:")
for item_id, combined_score in ranked_items:
    print(f"  {item_id}: combined_score={combined_score:.4f}")
````
