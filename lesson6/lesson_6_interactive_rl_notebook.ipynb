# Part 1: Static Models vs Adaptive Systems

A static model is trained once and then used for predictions.

Example:

```text
Train model → Use model → Predictions stay fixed
```

This is useful, but it has a limitation:

If a user changes their taste today, the model may not know until the next time it is retrained.

An adaptive system can respond to interaction:

```text
Show recommendation → User gives feedback → System updates → Better next recommendation
```

````python
# Static model example

user_preference = "Action"

def static_recommendation(preference):
    if preference == "Action":
        return "The Dark Knight"
    elif preference == "Comedy":
        return "The Grand Budapest Hotel"
    elif preference == "Sci-Fi":
        return "Interstellar"
    else:
        return "Finding Nemo"

print(static_recommendation(user_preference))
````

# Part 2: Feedback

Feedback tells the system how the user responded.

There are two main types of feedback:

## Explicit Feedback
The user directly tells the system what they think.

Examples:
- Like
- Dislike
- Rating
- Thumbs up / thumbs down

## Implicit Feedback
The system infers preference from behavior.

Examples:
- Watch time
- Clicks
- Skips
- Replays
- Purchases

````python
# Explicit feedback example

feedback = "like"

if feedback == "like":
    reward = 1
elif feedback == "dislike":
    reward = -1
else:
    reward = 0

print("Reward:", reward)
````

````python
# Implicit feedback example

clicked = 1
watch_percentage = 0.85

reward = clicked * watch_percentage

print("Reward:", reward)
````

# Part 4: Exploration vs Exploitation

A recommendation system has to balance two goals:

## Exploitation
Recommend what already works.

Example:
- The user likes action movies.
- Keep recommending action movies.

## Exploration
Try something new to learn more.

Example:
- The user likes action movies.
- Try one sci-fi action movie to see if they like it.

This is called the **Multi-Armed Bandit** problem.

````python
import random

known_favorites = ["Action Movie 1", "Action Movie 2", "Action Movie 3"]
new_options = ["Sci-Fi Movie", "Comedy Movie", "Documentary"]

exploration_rate = 0.3

if random.random() < exploration_rate:
    recommendation = random.choice(new_options)
    strategy = "Exploration"
else:
    recommendation = random.choice(known_favorites)
    strategy = "Exploitation"

print("Strategy:", strategy)
print("Recommendation:", recommendation)
````

# Part 6: Adaptive Scoring

Now we can use feature weights to score recommendations.

If an item belongs to a genre with a high weight, it should rank higher.

````python
items = [
    {"title": "The Dark Knight", "genre": "Action"},
    {"title": "Finding Nemo", "genre": "Animation"},
    {"title": "Interstellar", "genre": "Sci-Fi"},
    {"title": "The Grand Budapest Hotel", "genre": "Comedy"}
]

feature_weights = {
    "Action": 3,
    "Comedy": 0,
    "Sci-Fi": 2,
    "Animation": 1
}

for item in items:
    genre = item["genre"]
    item["score"] = feature_weights.get(genre, 0)

items_sorted = sorted(items, key=lambda x: x["score"], reverse=True)

for item in items_sorted:
    print(item["title"], "| Genre:", item["genre"], "| Score:", item["score"])
````

# Part 7: Simulating a Feedback Loop

Now we will simulate a small adaptive recommendation loop:

1. Show the top recommendation.
2. Receive feedback.
3. Update the feature weight.
4. Re-rank the items.
5. Show the next recommendation.

````python
items = [
    {"title": "The Dark Knight", "genre": "Action"},
    {"title": "Finding Nemo", "genre": "Animation"},
    {"title": "Interstellar", "genre": "Sci-Fi"},
    {"title": "The Grand Budapest Hotel", "genre": "Comedy"}
]

feature_weights = {
    "Action": 1,
    "Comedy": 1,
    "Sci-Fi": 1,
    "Animation": 1
}

def rank_items(items, weights):
    for item in items:
        item["score"] = weights.get(item["genre"], 0)
    return sorted(items, key=lambda x: x["score"], reverse=True)

def apply_feedback(genre, feedback, weights):
    if feedback == "like":
        weights[genre] += 1
    elif feedback == "dislike":
        weights[genre] -= 1
    return weights

# Round 1
ranked = rank_items(items, feature_weights)
top_item = ranked[0]

print("Recommended:", top_item["title"], "| Genre:", top_item["genre"])

# Simulated feedback
feedback = "like"
feature_weights = apply_feedback(top_item["genre"], feedback, feature_weights)

print("Updated weights:", feature_weights)

# Round 2
ranked = rank_items(items, feature_weights)

print("\nNew ranking:")
for item in ranked:
    print(item["title"], item["genre"], item["score"])
````
